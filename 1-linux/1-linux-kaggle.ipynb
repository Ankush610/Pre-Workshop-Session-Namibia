## 1. Where am I? 🧭

````python
!whoami
````

That answer is a **path**: the chain of folders from the top of the tree down to where you are.
Linux keeps everything in **one** big tree that starts at `/` (called *root*). Let's peek at the top of the tree with `ls` (**l**i**s**t):

````python
!ls /
````

These are system folders: `home` holds users' folders, `etc` holds settings, `bin` and `usr` hold programs, and `tmp` holds temporary files.

## 3. Creating files and listing them with `ls` 📁
What's in our new folder?

````python
!ls
````

**Nothing was printed, so the folder is empty.** Let's put some things in it:

| Command | What it does |
|---|---|
| `touch file` | create an empty file |
| `echo "text" > file` | create a file containing some text |
| `mkdir folder` | create a folder |

````python
!touch notes.txt                          # an empty file
!echo "Hello Linux!" > hello.txt          # a file with one line of text
!mkdir data scripts                       # two folders at once
!echo "name,score" > data/scores.csv      # a file INSIDE the data folder
````

Now `ls` has something to show:

````python
!ls
````

`ls` takes **flags** (options) that change what it shows. Let's try them one at a time.

**`-l`**: **l**ong format, which shows permissions, owner, size in bytes, and date

````python
!ls -l
````

- `notes.txt` is **0** bytes (empty, made by `touch`); `hello.txt` is **13** bytes (12 characters plus an invisible newline).
- Lines starting with **`d`** are **d**irectories (folders).

**`-a`**: show **a**ll files. Files whose names start with `.` are **hidden**, and plain `ls` skips them:

````python
!echo "you found me!" > .secret
!ls          # .secret is NOT shown
````

````python
!ls -a       # now it is. (. = this folder, .. = the parent folder)
````

**`-h`**: **h**uman-readable sizes. Let's make a bigger file to see the difference. `seq 1 200000` prints the numbers 1 to 200000:

````python
!seq 1 200000 > big.txt
!ls -l big.txt     # size in bytes: hard to read
!ls -lh big.txt    # size in K / M / G: easy to read
````

Give `ls` a folder name to look inside it, or use **`-R`** (**R**ecursive) to list everything inside every folder:

````python
!ls data
````

````python
!ls -R
````

Flags can be combined: `-lah` = `-l -a -h`. You'll use this one all the time:

````python
!ls -lah
````

💡 **Forgot a flag?** Almost every command has built-in help: `command --help` (or `man command` in a real terminal).

````python
!wc --help
````

## 4. Paths and moving around 🗺️
| Symbol | Meaning |
|---|---|
| `/` | root, the top of the tree |
| `~` | your home folder |
| `.` | the current folder |
| `..` | the parent folder (one level up) |

- An **absolute path** starts from `/` (or `~`) and works from anywhere: `~/playground/data`
- A **relative path** starts from where you are now: `data`

````python
!ls ~/playground/data    # absolute: works from anywhere
!ls data                 # relative: works because we're inside ~/playground
````

Let's walk into `data`, look around, then go back up with `..`:

````python
%cd data
!pwd
!ls ..       # list the parent folder, without moving
````

````python
%cd ..
!pwd
````

⚠️ **Notebook gotcha:** why `%cd` and not `!cd`? Each `!` line runs in its own short-lived shell, so the move is forgotten straight away. Watch:

````python
!cd data     # moves... inside a shell that closes immediately
!pwd         # still in ~/playground!
````

In a real terminal, plain `cd` works fine. In notebooks, use `%cd`.

## 5. Copying, moving, deleting ✂️
| Command | What it does |
|---|---|
| `cp src dst` | **c**o**p**y |
| `mv src dst` | **m**o**v**e, *or rename* |
| `rm file` | **r**e**m**ove a file |
| `rm -r folder` | remove a folder and everything inside it |

````python
!mkdir backup
!cp hello.txt backup/hello_copy.txt    # copy into another folder
!mv notes.txt data/                    # move into a folder
!mv hello.txt greeting.txt             # rename = "move" to a new name
!ls -R
````

⚠️ **`rm` has no recycle bin.** Deleted means gone forever, so always double-check before you press Enter.

````python
!rm big.txt       # delete a file
!rm -r backup     # delete a folder and everything inside it
!ls
````

## 7. Redirects and pipes 🔀
| Symbol | Meaning |
|---|---|
| `>` | send output into a file (**overwrites** it) |
| `>>` | send output into a file (**appends** to the end) |
| `\|` | **pipe**: feed the output of one command into the next one |

Pipes let you chain small tools into powerful one-liners.

````python
!echo "apple"  > fruits.txt
!echo "banana" >> fruits.txt
!echo "apple"  >> fruits.txt
!echo "cherry" >> fruits.txt
!cat fruits.txt
````

This cell uses `%%bash`, so every line runs as a normal shell command and no `!` is needed:

````python
%%bash
echo "--- sorted ---"
sort fruits.txt
echo "--- duplicates removed ---"
sort fruits.txt | uniq
echo "--- counted ---"
sort fruits.txt | uniq -c
````

Another pipe: how many things are in this folder?

````python
!ls
!ls | wc -l
````

## 8. Searching 🔍
**`grep`** prints the lines that contain some text.

| Flag | Meaning |
|---|---|
| `-i` | ignore upper/lower case |
| `-n` | show line numbers |
| `-c` | just **c**ount the matching lines |
| `-v` | in**v**ert: lines that do *not* match |

````python
%%bash
echo "--- lines containing 7 (first 5) ---"
grep 7 numbers.txt | head -5
echo "--- same, with line numbers ---"
grep -n 7 numbers.txt | head -5
echo "--- how many lines contain 7? ---"
grep -c 7 numbers.txt
````

**`find`** searches for files by name in a folder and every folder below it. `*` is a **wildcard** meaning "anything".

````python
!find . -name "*.txt"     # every .txt file, including the one inside data/
````

````python
!find . -name "*.csv"
````

````python
!find . -type d           # only directories (folders)
````

## 9. Permissions 🔐
In `ls -l` output, the first column looks like `-rwxr-xr-x`:
```
-   rwx    r-x    r-x
│   │      │      └── everyone else
│   │      └───────── group
│   └──────────────── owner (you)
└──────────────────── type: "-" = file, "d" = directory
```
`r` = read, `w` = write, `x` = execute (run it as a program), `-` = not allowed.

Let's write a tiny script and try to run it. Its first line, `#!/bin/bash` (a *shebang*), tells Linux which program should run the script.

````python
%%bash
cat > scripts/hello.sh << 'EOF'
#!/bin/bash
echo "Hi from my first script!"
echo "Today is $(date)"
EOF
ls -l scripts/hello.sh
````

````python
!./scripts/hello.sh    # ❌ Permission denied: there's no x (execute) flag yet
````

````python
!chmod +x scripts/hello.sh    # add execute permission
!ls -l scripts/hello.sh       # notice the x's now
!./scripts/hello.sh           # ./ means "starting from THIS folder"
````

## 11. Downloading data 🌐
- `wget URL` downloads a file (`-q` = quiet, `-O name` = save it as `name`)
- `curl -L -o name URL` does the same thing (common on Macs)

> **On Kaggle:** turn on **Settings → Internet** first, or downloads will fail.

We'll grab the famous **Titanic** passenger list into our `data` folder. We'll use it again in the Python and ML notebooks.

````python
!wget -q -O data/titanic.csv https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv
!ls -lh data
````

````python
!head -n 3 data/titanic.csv
````

Datasets often arrive zipped. `zip` packs files, `unzip` unpacks them.

````python
%%bash
zip -q data/titanic.zip data/titanic.csv     # pack
ls -lh data                                  # compare the sizes
unzip -o -q data/titanic.zip -d unzipped     # unpack into a new folder (-o = overwrite without asking)
ls -R unzipped
````
