## 6. Lists 📋
A list holds many values in order. Counting starts at **0**, and `-1` means the last item.

```python
# Many values in one box
temperatures = [20, 31, 28, 35, 30]
```

```python
print(temperatures[0])     # 31 - first item (counting starts at 0)
```

```python
print(temperatures[-1])    # 30 - last item
```

```python
print(len(temperatures))   # 4  - how many items
```

```python
temperatures.append(33)    # add to the end
print(temperatures)
```

```python
temperatures.insert(2, 20)     # add at a position 
print(temperatures)
```

```python
temperatures.extend([40, 42])  # add many items at once
print(temperatures)
```

```python
temperatures.remove(20)        # remove the first 20 found 
print(temperatures)
```

```python
val = temperatures.pop()             # remove the last item, gives it back
print(val)
print(temperatures)
```

```python
val = temperatures.pop(0)             # remove the item at position 0 
print(val)
print(temperatures)
```

```python
temperatures.sort()            # sort small to big  Note : It updated the original List 
print(temperatures)
```

```python
temperatures.sort(reverse=True)  # sort big to small
print(temperatures)
```

```python
temperatures.reverse()         # flip the order ( not in descending order just flipped it )
print(temperatures)
```

## Some more common functions

```python
print(temperatures)
print("len is :",len(temperatures))        # how many items      
print("max is :",max(temperatures))        # biggest             
print("min is :",min(temperatures))        # smallest          
print("sum is :",sum(temperatures))        # add them all up     
```

## 7. Dictionaries 📖
A dictionary stores **key → value** pairs, so you look things up by name instead of position.

```python
# Look up a value by its label (key)
phone = {"brand": "Samsung", 
         "price": 3500, 
         "colour": "blue"}
```

### Read

```python
print(phone["brand"])               # get a value by its key  -> "Samsung"
```

```python
print(phone["size"])  
```

```python
print(phone.get("size"))            # safe get: gives None instead of an error if the key is missing
```

```python
print(phone.get("size", "N/A"))     # safe get with a default  -> "N/A"
```

### Add

```python
# Change a value, or add a new one
phone["price"] = 3200
phone["storage"] = "128GB"
print(phone)
```

### Remove

```python
val = phone.pop("colour")          # remove a key, gives its value back -> "black"
print(val)
print(phone)
```

### Check

```python
val = "brand" in phone        # is this key in it?  -> True
print(val)
```

```python
# See everything
print(phone.keys())                 # all keys:   brand, price
print(phone.values())               # all values: Samsung, 3200
print(phone.items())                # key-value pairs
```

## 8. If / else 🤔
Let the program make decisions. Indentation (4 spaces) shows which lines belong to each branch.

```python
# Making decisions
temperature = 34
```

```python
if temperature > 30:
    print("It's hot, drink water")
else:
    print("Nice weather")
```

```python
# More than two choices with elif
if temperature > 38:
    print("Stay inside")
elif temperature > 25:
    print("Wear a hat")
else:
    print("Take a jacket")
```

```python
# = stores a value, == compares
print(temperature == 34)
```

```python
print(temperature != 34)
```

## 9. For loops 🔁
Repeat code for every item in a list, or a fixed number of times with `range()`.

```python
# Loop over a list
animals = ["lion", "elephant", "giraffe"]
for animal in animals:
    print("I saw a", animal)
```

```python
# Repeat a number of times (starts at 0, stops before 5)
for i in range(5):
    print("Step", i)
```

```python
# Add up all prices with a loop
prices = [20, 45, 15]
total = 0
for price in prices:
    total = total + price
print("Total:", total)
```

## 10. Functions 🛠️
`def` lets you write code once and reuse it. `return` sends the answer back.

```python
# Write once, use again
def square(number):
    return number * number
```

```python
print(square(3))
```

```python
print(square(7))
```

```python
# A function with two inputs
def welcome(name, course):
    return f"Welcome {name} to the {course} class!"
```

```python
print(welcome("Sam", "Python"))
```

## 12. ✍️ Exercise (5 min)
1. Make a list of 4 prices.
2. Use a `for` loop to print each price.
3. Write a function `add_vat(price)` that returns the price plus 15%.
4. Print the total of all prices **with** VAT.
5. ⭐ **Bonus:** print `"Expensive"` for any price over 100, otherwise `"Cheap"`.

```python
# your code here
```

```python
# @title 👀 Solution (try it yourself first!) { display-mode: "form" }
prices = [50, 120, 80, 200]

for price in prices:
    print(price)

def add_vat(price):
    return price * 1.15

total = 0
for price in prices:
    total = total + add_vat(price)
print("Total with VAT:", total)

for price in prices:
    if price > 100:
        print(price, "Expensive")
    else:
        print(price, "Cheap")
```
